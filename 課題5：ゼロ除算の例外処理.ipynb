```python
try:
    a = int(input("割られる数を入力してください:"))
    b = int(input("割る数を入力してください:"))

    c = a / b
    print(f"{a} ÷ {b} = {c}")

except ZeroDivisionError:
    print("エラー:0で割り算しないでください:")

except:
    print("エラー:数値を入力してください:")

finally:
    print("処理を終了します")
```

```python
try:
    a = int(input("割られる数を入力してください："))
    b = int(input("割る数を入力してください："))

    if b == 0:
        print("エラー:0で割り算しないでください:")

    else:
        c = a / b
    print(f"{a} ÷ {b} = {c}")

except ValueError:
    print("エラー：数値を入力してください:")

finally:
    print("処理を終了します")
```

```python
try:
    a = int(input("割られる数を入力してください："))
    b = int(input("割る数を入力してください："))

except ValueError:
    if b == 0:
        print("エラー：0で割り算しないでください:")

except:
    print("エラー：数値を入力してください:")

else:
    c = a / b
    print(f"{a} ÷ {b} = {c}")


finally:
    print("処理を終了します")
```
