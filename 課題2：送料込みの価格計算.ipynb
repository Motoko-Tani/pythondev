```python
price = int(input("税込価格を入力してください:"))
tax = round(price *1.1)

if tax >= 2000:
    print("送料は無料です")
    print("送料込みの価格は", tax, "円です。")
else:
    print("送料として350円かかります")
    payment = tax + 350
    print("送料込みの価格は",payment, "円です。")
```
