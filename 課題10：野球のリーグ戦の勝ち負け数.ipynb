```python
#勝ち負け数を管理するためのクラスを作成して、インスタンスを6チーム分生成した後に、各チームの勝ち負け数を表示する
#最初のセルに Baseball_Team という名前のクラスを作ってください
#name（チーム名が入ります：Archers／Meteors／Cheetahs／Sharks／Pythons／Owls）
#win（勝ち数）/lose（負け数）/draw（引き分け数）
#Baseball_Teamに勝率を算出する calc_win_rate() というメソッドを作ってください：
#勝率は、勝ち数/(勝ち数+負け数) にて算出しますときは、このcalc_win_rate()メソッドを実行します
#なお、今回は勝率の算出において、引き分け数は考慮しない形とします
#Baseball_Teamに、勝敗情報を表示する show_team_result() というメソッドを作ってください：
#勝敗情報の表示は、下記「出力結果の例」の表の1行分を表示するメソッドにしてください。例：Archers 77 64 2 0.546
#次のセルで Bshow_team_result() メソッドを呼び出して、勝敗情報を表示してください：
#書式指定を活用して、きれいな表を出力できていれば高評価としますが任意
#継承、多態性、カプセル化への対応も任意とします

class Baseball_Team:                              #Baseball_Teamクラスの定義
    def __init__(self, name, win, lose, draw):   #初期設定　コンストラクタ
        self.name = name                          #インスタンス変数の定義　このチームの名前を格納する変数
        self.win = win                                 
        self.lose = lose
        self.draw = draw 
       
    def calc_win_rate(self):                      #勝率を算出するメソッド    
        return self.win / (self.win + self.lose)

    def show_team_result(self):
        print(f"{self.name} {self.win} {self.lose} {self.draw} {self.calc_win_rate():.3f}")
```

```python
print("team win lose draw rate")                  #表の見出しを表示する

teams = [                                         #次のセルで Baseball_Team のインスタンスを6個チーム分、生成してください：
    Baseball_Team("Archers", 77, 64, 2),          #これらのインスタンスには、それぞれのチームの名前、勝ち数、負け数、引き分け数を格納します。
    Baseball_Team("Meteors", 80, 60, 3),
    Baseball_Team("Cheetahs", 75, 68, 4),
    Baseball_Team("Sharks", 82, 58, 3),
    Baseball_Team("Pythons", 70, 72, 5),
    Baseball_Team("Owls", 78, 66, 4),
]

for team in teams:
    team.show_team_result()
```
