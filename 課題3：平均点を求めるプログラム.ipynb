```python
total = 0           #合計点
people = 0          #人数

while True:         #無限ループ
    test = int(input("点数を入力してください:"))

    if test == -1:       #-1が出たら終わろう
        break
    total += test            #一人一人の点数を足していく（+=使う）
    people += 1              #人数も1人以上のはずだから足していくでしょ


if people == 0:
    print("テストの平均点は0.0点です")

else:
    avr = total / people     #平均値をとる
avr = round(avr,1)       #小数点を第1位までにする


print(f"{people}人のテストの平均点は{avr}点です")



```
