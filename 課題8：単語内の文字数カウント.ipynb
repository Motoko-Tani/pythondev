```python
english_word = {}                               #空の辞書を作る
for ch in "abcdefghijklmnopqrstuvwxyz":         #chにabc...を入れていく
    english_word[ch] = 0                        #key=abc...,value=0

word_list = []                                  #空の入れ物作る

while True:                                     #無限ループを作る
    word = input("英単語を入力してください:")     #英単語を入力してもらう
    if word == "":                              #何も入力されないときにstop
        break
    word_list.append(word)                      #入力した文字をword_listに保存していく

    for ch in word:                             #wordを1文字ずつchに入れていく
        if ch in english_word:                  #english_wordにchが入っている場合、回数を数える
            english_word[ch] += 1

for w in sorted(word_list):                     #wor_listをソートしてwに入れていく
    print(w)

for ch in "abcdefghijklmnopqrstuvwxyz":         #abc...をcxに入れていく
    if english_word[ch] > 0:                    #english_wordという辞書の中にabc...という文字が1以上あった場合出力
        print(f"{ch}が{english_word[ch]}個ありました")  #ch=key,english_word[0]=value

    
```
