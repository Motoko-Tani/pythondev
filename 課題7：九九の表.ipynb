```python
#@title 九九のリストを作成する
kuku = []                        #外側の箱を作る
for v in range(1, 10):
    gyo = []                     #横or縦に数字が入る
    for h in range(1, 10):
        gyo.append(v * h)        #縦×横をしていく　1×1...
    kuku.append(gyo)
print(kuku)
```

```python
#@title 九九のリストを作成する
kuku = []                        #外側の箱を作る
for v in range(1, 10):
    gyo = []                     #横or縦に数字が入る
    for h in range(1, 10):
        gyo.append(v * h)        #縦×横をしていく　1×1...
    kuku.append(gyo)

#@title 九九の表を表示する
for v in range(9):               #[1,2,3,...]は0～8dだから「9」にしないといけない
    for h in range(9):
        print(f"{kuku[v][h]:3d}", end="")     #上記で数字は出ているから*は使わないkukuという箱の中のv,hを使用

    print("")
```
