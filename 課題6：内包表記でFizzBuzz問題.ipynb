```python
def number_to_degree(a):             #関数作ってしまう
    if a % 3 == 0 and a % 5 == 0:    #まずは3と5で割り切れること
        return "FizzBuzz"
    else:                            #3で割り切れること
        if a % 3 == 0:
            return "Fizz"
        else:                        #5で割り切れること、だめなら数字を文字列で
            if a % 5 == 0:
                return "Buzz"
            else:
                return str(a)       #文字列はstrなはず

num6_alt = [number_to_degree(a) for a in range(1, 101)]    #, 101にすると0が入ってしまう
num6_alt
```
